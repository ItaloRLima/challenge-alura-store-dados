### Importação dos dados



```python
import pandas as pd
import numpy as np
import plotly.express as px

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

loja.head()
```

#1. Análise do faturamento



```python
faturamento1 = loja['Preço'].sum()
faturamento2 = loja2['Preço'].sum()
faturamento3 = loja3['Preço'].sum()
faturamento4 = loja4['Preço'].sum()
```

```python
print("Faturamento da loja 1: R$", faturamento1)
print("Faturamento da loja 2: R$", faturamento2)
print("Faturamento da loja 3: R$", faturamento3)
print("Faturamento da loja 4: R$", faturamento4)
```

```python
import matplotlib.pyplot as plt
lojas = ['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']
faturamentos = [faturamento1, faturamento2, faturamento3, faturamento4]

plt.ticklabel_format(style='plain', axis='y')
plt.bar(lojas,faturamentos,color=['tab:red','tab:brown','tab:green','tab:blue'],width=0.5)
# Loop para adicionar o texto em cada barra
for i, valor in enumerate(faturamentos):
    # (x, y, texto, alinhamento_horizontal, alinhamento_vertical)
    plt.text(i, valor, str(valor), ha='center', va='bottom')

plt.show()
```

# 2. Vendas por Categoria

```python
contagem1 = loja['Categoria do Produto'].value_counts()
contagem2 = loja2['Categoria do Produto'].value_counts()
contagem3 = loja3['Categoria do Produto'].value_counts()
contagem4 = loja4['Categoria do Produto'].value_counts()
```

```python
contagens = [contagem1, contagem2, contagem3, contagem4]

for x in contagens:
  print(x,'\n')
```

```python
print(contagem1.iloc[0:])
```

```python

#plt.title('Distribuição das categorias nas lojas.')
#plt.pie(contagem1.iloc[0:],labels=contagem1.index)
#plt.show()
fig, ((ax1,ax2),(ax3,ax4)) = plt.subplots(2,2,figsize=(15,10))
lista_axs = [ax1,ax2,ax3,ax4]
i=0

for ax in lista_axs:
  ax.set_title(f"Loja {i+1}")
  ax.pie(contagens[i].iloc[0:],labels=contagens[i].index,autopct='%1.1f%%')
  i+=1

plt.show()
```

# 3. Média de Avaliação das Lojas


```python
media1 = loja['Avaliação da compra'].mean()
media2 = loja2['Avaliação da compra'].mean()
media3 = loja3['Avaliação da compra'].mean()
media4 = loja4['Avaliação da compra'].mean()
```

```python
medias = [media1,media2,media3,media4]
for i, valor in enumerate(medias):
  print(f"Média da loja {i+1}: {valor:.2f}")
```

```python
plt.ticklabel_format(style='plain', axis='y')
plt.scatter(lojas,medias,color=['tab:red','tab:brown','tab:green','tab:blue'])
plt.title('Avaliação média das lojas')

for i, valor in enumerate(medias):
    # (x, y, texto, alinhamento_horizontal, alinhamento_vertical)
    plt.text(i, valor, f"{valor:.2f}", ha='center', va='bottom')

plt.show()
```

# 4. Produtos Mais e Menos Vendidos

```python
contagemProdl1 = loja['Produto'].value_counts()
contagemProdl2 = loja2['Produto'].value_counts()
contagemProdl3 = loja3['Produto'].value_counts()
contagemProdl4 = loja4['Produto'].value_counts()

contagensProdls = [contagemProdl1,contagemProdl2,contagemProdl3,contagemProdl4]
```

```python
for x, produto in enumerate(contagensProdls):
  print(f"O produto mais vendido na loja {x+1} foi {produto.index[0].upper()}, e o menos vendido foi {produto.index[-1].upper()}")
```

```python
produtosMaisVendidos = []
produtosMenosVendidos = []

for i in contagensProdls:
  for x in range(2):
    produtosMaisVendidos.append([i.index[x],int(i.iloc[x])])
  for w in range(-2,0,1):
    produtosMenosVendidos.append([i.index[w],int(i.iloc[w])])

print(produtosMaisVendidos,"\n",produtosMenosVendidos)
```

# 5. Frete Médio por Loja

```python
loja1Frete = loja['Frete'].mean()
loja2Frete = loja2['Frete'].mean()
loja3Frete = loja3['Frete'].mean()
loja4Frete = loja4['Frete'].mean()

fretes = [loja1Frete,loja2Frete,loja3Frete,loja4Frete]
```

```python
plt.ticklabel_format(style='plain', axis='y')
plt.bar(lojas,fretes,width=0.2,color=['tab:red','tab:blue','tab:orange','tab:green'])
plt.title('Preço médio gasto com frete em cada loja')

for i, valor in enumerate(fretes):
    # (x, y, texto, alinhamento_horizontal, alinhamento_vertical)
    plt.text(i, valor, f"{valor:.2f}", ha='center', va='bottom')

plt.show()
```

```python
tds_lojas = pd.concat([loja,loja2,loja3,loja4])

loja_agrupada = tds_lojas.groupby(['lat', 'lon']).size().reset_index(name='Quantidade')

fig = px.density_map(loja_agrupada, lat='lat', lon='lon', z = 'Quantidade', radius=20,
                        center=dict(lat=-15, lon=-49), zoom=2.7,
                        map_style="open-street-map",title='Distribuição geográfica dos produtos vendidos')
fig.update_layout(
    title_x=0.5,
    title_font_size=20
)
fig.show()
```
